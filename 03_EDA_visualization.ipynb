## 9. Rating Analysis

Rating analysis is performed to understand how customers have
rated the products.

The rating values range from 1 to 5, where higher ratings indicate
a more positive customer evaluation.

This analysis helps in understanding the overall customer
satisfaction pattern in the dataset.

```python
rating_counts = df["Rate"].value_counts().sort_index()

print("Customer Rating Distrabution:")
display(rating_counts)
```

## 9.1 Rating Distribution Visualization

A bar chart is used to visualize the distribution of customer
ratings from 1 to 5.

This visualization makes it easier to compare the number of
reviews received for each rating.

```python
rating_counts = df["Rate"].value_counts().sort_index()

plt.figure(figsize=(8, 5))

plt.bar(
    rating_counts.index.astype(str),
    rating_counts.values
)

plt.title("Customer Rating Distribution")
plt.xlabel("Rating")
plt.ylabel("Number of Reviews")

plt.tight_layout()
plt.show()
```

## 9.2 Rating Analysis Result

The rating distribution shows that 5-star ratings have the highest
number of customer reviews, followed by 4-star ratings.

Lower ratings such as 1-star and 2-star reviews are comparatively
less frequent.

This indicates that the dataset contains a larger proportion of
positive customer feedback.

## 10. Product Analysis

Product analysis is performed to understand the distribution of
customer reviews across different products.

This analysis helps identify the products that receive a higher
volume of customer feedback.

```python
product_counts = df["Product_name"].value_counts()

print("Number of Unique Products:", df["Product_name"].nunique())

print("\nTop 10 Products:")
display(product_counts.head(10))
```

## 10.1 Top 10 Most Reviewed Products

The top 10 products with the highest number of customer reviews
are identified to understand which products receive the most
customer feedback.

A bar chart is used to compare the review counts of these products.

```python
top_products = df["Product_name"].value_counts().head(10)

plt.figure(figsize=(12, 7))

plt.barh(
    top_products.index[::-1],
    top_products.values[::-1]
)

plt.title("Top 10 Most Reviewed Products")
plt.xlabel("Number of Reviews")
plt.ylabel("Product")

plt.tight_layout()
plt.show()
```

## 10.2 Product Analysis Result

The dataset contains 1,161 unique products.

The analysis shows that some products receive a significantly higher
number of customer reviews than others.

The most reviewed products include the cello Dinner Set, Bumtum
Baby Diaper Pants, Google Home Mini, and Mi Smart TV.

This indicates that customer feedback is concentrated around
certain products, which can help identify products with high
customer engagement.

## 11. Review Statistics

Review statistics are analyzed to understand the length and
characteristics of customer reviews.

Review length helps identify whether customers generally provide
short or detailed feedback.

```python
df["Review_Length"] = df["Review"].fillna("").str.len()

print("Review Length Statistics:")
display(df["Review_Length"].describe())
```

## 11.1 Review Length Distribution

A histogram is used to visualize the distribution of review lengths.

This visualization helps understand whether customer reviews are
generally short or detailed and identifies the overall pattern of
review length.

```python
plt.figure(figsize=(10,5))

plt.hist(
    df["Review_Length"],
    bins=50
)

plt.title("Distrabution of Review Length")
plt.xlabel("Review Length(Characters)")
plt.ylabel("Number of Reviwes")

plt.tight_layout()
plt.show()
```

## 11.2 Review Statistics Result

The review length analysis shows that customer reviews are generally short.

The average review length is approximately 12.71 characters, while the
median review length is 11 characters. The maximum review length is
358 characters.

This indicates that most customers provide short feedback, while a
smaller number of customers provide longer and more detailed reviews.

## 12. Price Analysis

Price analysis is performed to understand the price distribution
of products in the customer feedback dataset.

This analysis helps identify the typical price range of products
and understand how product prices are distributed.

```python
print("Price statistics:")

print("Minimam Price: ₹", df["Price"].min())
print("Maximam Price: ₹", df["Price"].max())
print("Average Price: ₹", round(df["Price"].mean(), 2))
print("Median Price:  ₹", df["Price"].median())
```

## 12.1 Price Distribution Visualization

A histogram is used to visualize the distribution of product prices.

This visualization helps identify the common price range of products
and understand how prices are distributed across the dataset.

```python
plt.figure(figsize=(10, 5))

plt.hist(
    df["Price"],
    bins=50
)

plt.title("Distribution of Product Prices")
plt.xlabel("Price (₹)")
plt.ylabel("Number of Products")

plt.tight_layout()
plt.show()
```

## 12.2 Price Analysis Result

The price distribution shows that most products are concentrated
in the lower price range.

The average product price is approximately ₹4,755.55, while the
median price is ₹1,999. This difference indicates that a smaller
number of high-priced products increase the overall average price.

The maximum product price in the dataset is ₹86,990, showing a
wide variation in product prices.

## 13. Summary Statistics

The summary text is analyzed to understand the length of
customer-provided summaries.

This analysis helps identify whether customers generally provide
short summaries or more detailed feedback.

```python
df["Summary_Length"] = df["Summary"].fillna("").str.len()

print("Summary Length Statistics:")
display(df["Summary_Length"].describe())
```

## 13.1 Summary Length Distribution

A histogram is used to visualize the distribution of summary lengths.

This visualization helps understand whether customer summaries are
generally short or detailed and shows the variation in summary length.

```python
plt.figure(figsize=(10, 5))

plt.hist(
    df["Summary_Length"],
    bins=50
)

plt.title("Distribution of Summary Length")
plt.xlabel("Summary Length (Characters)")
plt.ylabel("Number of Summaries")

plt.tight_layout()
plt.show()
```

## 13.2 Summary Analysis Result

The summary length analysis shows that customer summaries are generally
short in length.

The average summary length is approximately 51.24 characters, while
the median length is 23 characters. The maximum summary length is
940 characters.

This indicates that most customers provide short summaries, while
some customers provide longer and more detailed summaries.

## 14. Rating vs Price Analysis

The relationship between product price and customer rating is
analyzed to understand whether product price has any visible
association with customer ratings.

A scatter plot is used to visualize the relationship between
Price and Rate.

```python
plt.figure(figsize=(10, 6))

plt.scatter(
    df["Price"],
    df["Rate"],
    alpha=0.3
)

plt.title("Price vs Customer Rating")
plt.xlabel("Price (₹)")
plt.ylabel("Rating")

plt.tight_layout()
plt.show()
```
