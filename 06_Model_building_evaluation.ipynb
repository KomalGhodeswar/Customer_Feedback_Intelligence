# Customer Feedback Intelligence System

## Day 7 & Day 8 — Model Building and Evaluation

**Project Type:** Data Science / Machine Learning / NLP  

**Domain:** Customer Analytics, NLP, Machine Learning

## 1. Objective

The objective of this task is to build and evaluate machine learning
models for customer sentiment classification.

In Day 7, machine learning classification models will be trained using
the TF-IDF features generated from customer feedback text.

In Day 8, the trained models will be evaluated using Accuracy,
Precision, Recall, F1-Score, and Confusion Matrix to identify the
best-performing model for customer feedback sentiment analysis.

The selected model will be used as the foundation for further
sentiment analysis and customer feedback intelligence.

```python
import pandas as pd
import numpy as np

from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import MultinomialNB

from sklearn.model_selection import train_test_split

from sklearn.feature_extraction.text import TfidfVectorizer

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

print("Libraries imported successfully!")
```

## 2. Load Cleaned Dataset

The cleaned customer feedback dataset prepared during the previous
stages is loaded for model building and evaluation.

The dataset contains the processed `Review` and `Summary` text along
with the `Rate` column, which will be used to create sentiment labels.

```python
cleaned_file_path = r"C:\archive\Customer_Feedback_Cleaned.csv"

df = pd.read_csv(
    cleaned_file_path,
    encoding="utf-8"
)

print("Cleaned dataset loaded successfully!")
print("Dataset Shape:", df.shape)
print("Columns:", df.columns.tolist())
```

## 3. Create Sentiment Labels

The `Rate` column is used to create sentiment labels for customer
feedback.

Ratings of 4 and 5 are classified as **Positive**, rating 3 as
**Neutral**, and ratings of 1 and 2 as **Negative**.

The `Sentiment` column will be used as the target variable for
machine learning classification.

```python
def create_sentiment(rating):
    if rating >= 4:
        return "Positive"
    elif rating == 3:
        return "Neutral"
    else:
        return "Negative"


df["Sentiment"] = df["Rate"].apply(create_sentiment)

print("Sentiment labels created successfully!")

print("\nSentiment Distribution:")
print(df["Sentiment"].value_counts())

display(
    df[["Rate", "Sentiment"]].head(10)
)
```

## 4. Prepare Text Features

The processed `Review` and `Summary` text are combined into a single
text feature.

This combined text will be used as the input feature for TF-IDF
vectorization and subsequent machine learning model training.

```python


df["Combined_Text"] = (
    df["Review"].fillna("") + " " + df["Summary"].fillna("")
)

print("Review and Summary combined successfully!")

display(
    df[["Review", "Summary", "Combined_Text"]].head(5)
)
```

## 5. Train-Test Split

The combined customer feedback text and sentiment labels are divided
into training and testing datasets.

The training data will be used to train the machine learning models,
while the testing data will be used to evaluate their performance on
unseen customer feedback.

An 80:20 split is used for training and testing.

```python

X = df["Combined_Text"]
y = df["Sentiment"]


X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Train-Test split completed successfully!")

print("\nTraining samples:", len(X_train))
print("Testing samples:", len(X_test))

print("\nTraining target distribution:")
print(y_train.value_counts())

print("\nTesting target distribution:")
print(y_test.value_counts())
```

## 6. TF-IDF Feature Extraction

The combined customer feedback text is converted into numerical
features using TF-IDF.

Unigrams and bigrams are used to capture both individual words and
two-word combinations from customer feedback.

The TF-IDF vectorizer is fitted only on the training data and then
used to transform the testing data to avoid data leakage.

```python

tfidf_vectorizer = TfidfVectorizer(
    ngram_range=(1, 2),
    max_features=5000,
    min_df=2
)


X_train_tfidf = tfidf_vectorizer.fit_transform(X_train)


X_test_tfidf = tfidf_vectorizer.transform(X_test)

print("TF-IDF feature extraction completed successfully!")

print("\nTraining TF-IDF Shape:", X_train_tfidf.shape)
print("Testing TF-IDF Shape:", X_test_tfidf.shape)
```

# Model Building

Logistic Regression is used as the first classification model for
customer sentiment prediction.

The model is trained using the TF-IDF features generated from
customer feedback text.

The model learns the relationship between the text features and the
sentiment labels: Positive, Neutral, and Negative.

```python
logistic_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

logistic_model.fit(
    X_train_tfidf,
    y_train
)

print("Logistic Regression model trained successfully!")
```

## 7. Multinomial Naive Bayes

Multinomial Naive Bayes is a classification algorithm that is
commonly used for text classification problems.

In this step, the model is trained using the TF-IDF features and
sentiment labels.

The model learns patterns from the customer feedback text to
classify reviews into Positive, Neutral, and Negative sentiment.

```python
naive_bayes_model = MultinomialNB()

naive_bayes_model.fit(
    X_train_tfidf,
    y_train
)

print("Multinomial Naive Bayes trained successfully!")
```

#  8 — Model Evaluation

The trained Logistic Regression and Multinomial Naive Bayes models
are used to predict the sentiment of unseen customer feedback from
the testing dataset.

The predictions will be compared with the actual sentiment labels
to evaluate the performance of both models.

```python
y_pred_logistic = logistic_model.predict(X_test_tfidf)

y_pred_naive_bayes = naive_bayes_model.predict(X_test_tfidf)

print("Predictions generated successfully!")

print("\nLogistic Regression - Sample Predictions:")
print(y_pred_logistic[:10])

print("\nMultinomial Naive Bayes - Sample Predictions:")
print(y_pred_naive_bayes[:10])
```

## 9. Accuracy

Accuracy measures the overall percentage of correct predictions made
by a classification model.

It is calculated by comparing the predicted sentiment labels with
the actual sentiment labels from the testing dataset.

Accuracy will be calculated for both Logistic Regression and
Multinomial Naive Bayes models.

```python
logistic_accuracy = accuracy_score(
    y_test,
    y_pred_logistic
)

print("Model Acuuracy:")

print(
    "Logistic Regression:",
    round(logistic_accuracy,4)
)
print(
    "Multinominal Naive Bayes:",
     round(naive_bayes_accuracy,4)
)
```

## 10. Precision

Precision measures how many of the sentiment predictions made by a
model were actually correct.

In this step, weighted Precision is calculated for both Logistic
Regression and Multinomial Naive Bayes models.

A higher Precision indicates that the model makes fewer incorrect
sentiment predictions.

```python
logistic_precision = precision_score(
    y_test,
    y_pred_logistic,
    average="weighted"
)

naive_bayes_precision = precision_score(
    y_test,
    y_pred_naive_bayes,
    average="weighted"
)

print("Model Precision:")

print(
    "Logistic Regression:",
    round(logistic_precision,4)
)

print(
    "Multinomial Naive Bayes:",
    round(naive_bayes_precision,4)
)
```

## 11. Recall

Recall measures how many of the actual sentiment cases were correctly
identified by the model.

In this step, weighted Recall is calculated for both Logistic
Regression and Multinomial Naive Bayes models.

A higher Recall indicates that the model is better at identifying
the actual sentiment classes.

```python


logistic_recall = recall_score(
    y_test,
    y_pred_logistic,
    average="weighted"
)

naive_bayes_recall = recall_score(
    y_test,
    y_pred_naive_bayes,
    average="weighted"
)

print("Model Recall:")

print(
    "Logistic Regression:",
    round(logistic_recall, 4)
)

print(
    "Multinomial Naive Bayes:",
    round(naive_bayes_recall, 4)
)
```

## 12. F1-Score

F1-Score provides a balanced measure of Precision and Recall.

It is useful when both correct predictions and the identification
of actual sentiment cases are important.

In this step, weighted F1-Score is calculated for both Logistic
Regression and Multinomial Naive Bayes models.

```python


logistic_f1 = f1_score(
    y_test,
    y_pred_logistic,
    average="weighted"
)

naive_bayes_f1 = f1_score(
    y_test,
    y_pred_naive_bayes,
    average="weighted"
)

print("Model F1-Score:")

print(
    "Logistic Regression:",
    round(logistic_f1, 4)
)

print(
    "Multinomial Naive Bayes:",
    round(naive_bayes_f1, 4)
)
```

## 13. Confusion Matrix

A Confusion Matrix is used to examine the classification results of
a machine learning model.

It shows the relationship between the actual sentiment labels and
the predicted sentiment labels.

In this step, Confusion Matrices will be generated for both Logistic
Regression and Multinomial Naive Bayes models.

```python


logistic_cm = confusion_matrix(
    y_test,
    y_pred_logistic,
    labels=["Negative", "Neutral", "Positive"]
)

naive_bayes_cm = confusion_matrix(
    y_test,
    y_pred_naive_bayes,
    labels=["Negative", "Neutral", "Positive"]
)

print("Logistic Regression - Confusion Matrix:")
print(logistic_cm)

print("\nMultinomial Naive Bayes - Confusion Matrix:")
print(naive_bayes_cm)
```

## 14. Model Comparison

The performance of Logistic Regression and Multinomial Naive Bayes
models is compared using Accuracy, Precision, Recall, and F1-Score.

This comparison helps identify which model performs better for
customer sentiment classification.

```python


model_comparison = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Multinomial Naive Bayes"
    ],
    "Accuracy": [
        logistic_accuracy,
        naive_bayes_accuracy
    ],
    "Precision": [
        logistic_precision,
        naive_bayes_precision
    ],
    "Recall": [
        logistic_recall,
        naive_bayes_recall
    ],
    "F1-Score": [
        logistic_f1,
        naive_bayes_f1
    ]
})


model_comparison[
    ["Accuracy", "Precision", "Recall", "F1-Score"]
] = model_comparison[
    ["Accuracy", "Precision", "Recall", "F1-Score"]
].round(4)

display(model_comparison)
```

## 14. Best Model Selection

The best-performing model is selected based on the evaluation
metrics obtained during model evaluation.

F1-Score is given higher importance because it provides a balanced
measure of Precision and Recall.

The model with the highest F1-Score is selected as the best model
for customer sentiment classification.

```python


best_model_row = model_comparison.loc[
    model_comparison["F1-Score"].idxmax()
]

best_model_name = best_model_row["Model"]
best_f1_score = best_model_row["F1-Score"]

print("Best Model:", best_model_name)
print("Best F1-Score:", best_f1_score)
```

## 16. Conclusion

The model building and evaluation stages were successfully completed.

Two machine learning classification models, Logistic Regression and
Multinomial Naive Bayes, were trained using TF-IDF features generated
from customer feedback text.

The models were evaluated using Accuracy, Precision, Recall,
F1-Score, and Confusion Matrix.

The model with the highest F1-Score was selected as the best-performing
model for customer sentiment classification.

The selected model can be used in the next stages of the project for
sentiment analysis and customer feedback intelligence.
