## 6. Check Missing Values

Missing values are checked in all columns before performing
any data cleaning operations.

```python
print("Missing Values:")
display(df.isnull().sum())
```

## 6.1 Missing Value Percentage

The percentage of missing values is calculated for each column to
understand the extent of missing data in the dataset.

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

print("Missing Value Percentage:")
display(missing_percentage)
```

## 7. Check Duplicate Records

Duplicate records are checked to identify repeated rows in the
dataset before further cleaning.

```python
duplicate_count = df.duplicated().sum()

print("Number of Duplicate Rows:", duplicate_count)
```

## 8. Remove Duplicate Records

Duplicate records are removed to avoid repeated data from affecting
the results of further analysis and machine learning tasks.

```python
rows_before = len(df)

df = df.drop_duplicates()

rows_after = len(df)

print("Rows Before Removing Duplicates:", rows_before)
print("Rows After Removing Duplicates:", rows_after)
print("Duplicate Rows Removed:", rows_before - rows_after)
```

## 9. Price Column Cleaning

The Price column contains currency symbols and unwanted characters.
These characters are removed and the column is converted into a
numerical format for further analysis.

```python


df["Price"] = (
    df["Price"]
    .astype(str)
    .str.replace(r"[^\d.]", "", regex=True)
)

df["Price"] = pd.to_numeric(df["Price"], errors="coerce")

print("Price Data Type After Cleaning:", df["Price"].dtype)

print("\nSample Price Values:")
for price in df["Price"].head(10):
    print(f"₹ {price:,.1f}")
```

## 10. Rate Column Cleaning

The Rate column is converted into a numerical format so that
rating-based analysis and visualization can be performed.

```python


df["Rate"] = pd.to_numeric(df["Rate"], errors="coerce")

print("Rate Data Type After Cleaning:", df["Rate"].dtype)

print("\nUnique Rating Values:")
print(df["Rate"].value_counts().sort_index())
```

## 11. Handle Missing Values

Missing values are handled based on the type and importance of
each column.

Missing product names are replaced with "Unknown Product".
Missing Price values are replaced with the median Price.
Rows with missing Rate values are removed because Rating is an
important variable for sentiment and rating-based analysis.

Missing Review and Summary values are replaced with empty strings
to make the text data suitable for NLP preprocessing.

```python

df["Product_name"] = df["Product_name"].fillna("Unknown Product")


df["Price"] = df["Price"].fillna(df["Price"].median())


rows_before_rate = len(df)
df = df.dropna(subset=["Rate"])
rows_after_rate = len(df)


df["Review"] = df["Review"].fillna("")
df["Summary"] = df["Summary"].fillna("")

print("Missing values handled successfully!")
print("Rows removed due to missing Rate:", rows_before_rate - rows_after_rate)
```

## 12. Final Missing Value Verification

Missing values are checked again after the cleaning process to
confirm that the dataset is properly prepared for further analysis.

```python
print("Final Missing Values:")
display(df.isnull().sum())
```

## 13. Final Dataset Verification

The cleaned dataset is verified by checking its shape, column names,
data types, and sample records before saving the final dataset.

```python
print("Final Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nFirst 5 Records:")
display(df.head())
```

## 14. Product Name Text Cleaning

The Product_name column is cleaned by removing unwanted encoding
characters while preserving the actual product name.

```python


df["Product_name"] = (
    df["Product_name"]
    .astype(str)
    .str.replace(r"[?]", "", regex=True)
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)

print("Product Name cleaned successfully!")

print("\nSample Product Names:")
display(df["Product_name"].head())
```

## 15. Save Cleaned Dataset

The cleaned customer feedback dataset is saved as a new CSV file
for use in Exploratory Data Analysis, Visualization, NLP
preprocessing, and Machine Learning.

```python
cleaned_file_path = r"C:\archive\Customer_Feedback_Cleaned.csv"

df.to_csv(
    cleaned_file_path,
    index=False,
    encoding="utf-8"
)

print("Cleaned dataset saved successfully!")
print("File:", cleaned_file_path)
```
