## 1. Objective

The objective of this task is to preprocess customer feedback text
for Natural Language Processing (NLP).

In this task, the `Review` and `Summary` columns will be processed
using text cleaning, tokenization, stopword removal, stemming,
and lemmatization.

The processed text will be used in the next stages for feature
engineering and sentiment analysis.

```python
import pandas as pd
import re 
import nltk

print("Libraries imported successfully!")
```

```python
cleaned_file_path = r"C:\archive\Customer_Feedback_Cleaned.csv"

df = pd.read_csv(cleaned_file_path, encoding="utf-8")

print("Cleaned dataset loaded successfully!")
print("Dataset Shape:", df.shape)
print("Columns:", df.columns.tolist())
```

## 2. Inspect Review and Summary Text

The `Review` and `Summary` columns contain customer feedback text.

Before applying NLP preprocessing, sample records are inspected to
understand the structure and content of the text data.

```python
print("Sample Customer Reviews:")
display(df["Review"].head(10))

print("\nSample Customer Summaries:")
display(df["Summary"].head(10))
```

## 3. Text Cleaning

Text cleaning is performed to remove unnecessary characters and
standardize the customer feedback text.

The `Review` and `Summary` columns are converted to lowercase,
special characters and numbers are removed, and extra spaces are
eliminated.

This step prepares the text for further NLP processing.

```python
def clean_text(text):
    text = str(text).lower()
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text


df["Cleaned_Review"] = df["Review"].apply(clean_text)
df["Cleaned_Summary"] = df["Summary"].apply(clean_text)

print("Text cleaning completed successfully!")

display(
    df[["Review", "Cleaned_Review", "Summary", "Cleaned_Summary"]].head(10)
)
```

## 4. Tokenization

Tokenization is the process of splitting text into individual words
or tokens.

In this step, the cleaned `Review` and `Summary` text will be
converted into lists of individual words.

Tokenization is an important step because NLP models work with
individual words or tokens during further text processing.

```python
from nltk.tokenize import word_tokenize

nltk.download("punkt")
nltk.download("punkt_tab")

df["Review_Tokens"] = df["Cleaned_Review"].apply(word_tokenize)
df["Summary_Tokens"] = df["Cleaned_Summary"].apply(word_tokenize)

print("Tokenization completed successfully!")

display(
    df[
        [
            "Cleaned_Review",
            "Review_Tokens",
            "Cleaned_Summary",
            "Summary_Tokens"
        ]
    ].head(10)
)
```

## 5. Stopword Removal

Stopwords are common words such as `the`, `is`, `and`, `of`, and `to`
that usually provide limited meaning for text analysis.

In this step, stopwords will be removed from the tokenized `Review`
and `Summary` text.

This helps reduce unnecessary words and keeps the more meaningful
terms for further NLP processing.

```python
from nltk.corpus import stopwords

nltk.download("stopwords")

stop_words = set(stopwords.words("english"))

df["Filtered_Review_Tokens"] = df["Review_Tokens"].apply(
    lambda tokens: [word for word in tokens if word not in stop_words]
)

df["Filtered_Summary_Tokens"] = df["Summary_Tokens"].apply(
    lambda tokens: [word for word in tokens if word not in stop_words]
)

print("Stopword removal completed successfully!")

display(
    df[
        [
            "Review_Tokens",
            "Filtered_Review_Tokens",
            "Summary_Tokens",
            "Filtered_Summary_Tokens"
        ]
    ].head(10)
)
```

## 6. Stemming

Stemming is the process of reducing words to their root or base form
by removing word endings.

In this step, the Porter Stemmer will be applied to the filtered
tokens from both the `Review` and `Summary` columns.

Stemming helps reduce different forms of a word to a common form,
which can reduce the number of unique terms during text analysis.

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

df["Stemmed_Review_Tokens"] = df["Filtered_Review_Tokens"].apply(
    lambda tokens: [stemmer.stem(word) for word in tokens]
)

df["Stemmed_Summary_Tokens"] = df["Filtered_Summary_Tokens"].apply(
    lambda tokens: [stemmer.stem(word) for word in tokens]
)

print("Stemming completed successfully!")

display(
    df[
        [
            "Filtered_Review_Tokens",
            "Stemmed_Review_Tokens",
            "Filtered_Summary_Tokens",
            "Stemmed_Summary_Tokens"
        ]
    ].head(10)
)
```

## 7. Lemmatization

Lemmatization is the process of converting words into their meaningful
base or dictionary form.

Unlike stemming, lemmatization produces valid words by considering
the actual language structure.

In this step, lemmatization will be applied to the filtered tokens
from both the `Review` and `Summary` columns.

```python
from nltk.stem import WordNetLemmatizer

nltk.download("wordnet")
nltk.download("omw-1.4")

lemmatizer = WordNetLemmatizer()

df["Lemmatized_Review_Tokens"] = df["Filtered_Review_Tokens"].apply(
    lambda tokens: [lemmatizer.lemmatize(word) for word in tokens]
)

df["Lemmatized_Summary_Tokens"] = df["Filtered_Summary_Tokens"].apply(
    lambda tokens: [lemmatizer.lemmatize(word) for word in tokens]
)

print("Lemmatization completed successfully!")

display(
    df[
        [
            "Filtered_Review_Tokens",
            "Lemmatized_Review_Tokens",
            "Filtered_Summary_Tokens",
            "Lemmatized_Summary_Tokens"
        ]
    ].head(10)
)
```

## 8. Create Final Processed Text

The lemmatized tokens are converted back into text format to create
the final processed text.

The processed `Review` and `Summary` texts will be used in the next
stage for feature engineering and TF-IDF feature extraction.

```python
df["Processed_Review"] = df["Lemmatized_Review_Tokens"].apply(
    lambda tokens: " ".join(tokens)
)

df["Processed_Summary"] = df["Lemmatized_Summary_Tokens"].apply(
    lambda tokens: " ".join(tokens)
)

print("Final processed text created successfully!")

display(
    df[
        [
            "Review",
            "Processed_Review",
            "Summary",
            "Processed_Summary"
        ]
    ].head(10)
)
```

## 9. Save the Cleaned Dataset

The NLP preprocessing steps have been completed for both the `Review`
and `Summary` columns.

The intermediate NLP columns are temporary and are not required in
the final cleaned dataset. Therefore, only the original five dataset
columns are retained and saved.

The original `Dataset.csv` remains unchanged.

```python
final_columns = [
    "Product_name",
    "Price",
    "Rate",
    "Processed_Review",
    "Processed_Summary"
]

df_final = df[final_columns].copy()

df_final = df_final.rename(
    columns={
        "Processed_Review": "Review",
        "Processed_Summary": "Summary"
    }
)

df_final.to_csv(
    cleaned_file_path,
    index=False,
    encoding="utf-8"
)

print("NLP processed dataset saved successfully!")
print("Final Shape:", df_final.shape)
print("Final Columns:", df_final.columns.tolist())
```

The final cleaned dataset is verified to ensure that the required
